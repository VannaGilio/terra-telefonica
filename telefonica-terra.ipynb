```python
import pandas as pd

servicos = [
    "ServicoTelefonico",
    "MultiplasLinhas",
    "ServicoInternet",
    "SegurancaOnline",
    "BackupOnline",
    "ProtecaoDispositivo",
    "SuporteTecnico",
    "StreamingTV",
    "StreamingFilmes",
]

nomes = {
    "ServicoTelefonico": "Telefone",
    "MultiplasLinhas": "Múltiplas linhas",
    "ServicoInternet": "Internet",
    "SegurancaOnline": "Segurança online",
    "BackupOnline": "Backup online",
    "ProtecaoDispositivo": "Proteção do dispositivo",
    "SuporteTecnico": "Suporte técnico",
    "StreamingTV": "Streaming de TV",
    "StreamingFilmes": "Streaming de filmes",
}

cancelou = (
    df_gold["Cancelamento"]
    .astype("string")
    .str.strip()
    .str.casefold()
    .isin(["sim", "yes", "true", "1"])
)
cancelados = df_gold.loc[cancelou]

positivos = {"sim", "yes", "true", "1"}
valores_sem_internet = {
    "não", "nao", "no", "false", "0",
    "no internet service", ""
}

contagem = {}

for coluna in servicos:
    valores = (
        cancelados[coluna]
        .astype("string")
        .str.strip()
        .str.casefold()
    )

    if coluna == "ServicoInternet":
        tinha_servico = valores.notna() & ~valores.isin(valores_sem_internet)
    else:
        tinha_servico = valores.isin(positivos)

    contagem[nomes[coluna]] = int(tinha_servico.sum())

dados_grafico = (
    pd.Series(contagem, name="Clientes")
    .rename_axis("Serviço")
    .reset_index()
    .sort_values("Clientes", ascending=True)
)

total_cancelados = len(cancelados)
dados_grafico["Percentual"] = (
    dados_grafico["Clientes"] / total_cancelados * 100
    if total_cancelados > 0
    else 0
)

fig = px.bar(
    dados_grafico,
    x="Clientes",
    y="Serviço",
    orientation="h",
    color="Clientes",
    text="Clientes",
    custom_data=["Percentual"],
    color_continuous_scale="Plasma",
    title="Serviços entre clientes que cancelaram",
    labels={
        "Clientes": "Quantidade de clientes",
        "Serviço": "",
    },
    template="plotly_white",
)

fig.update_traces(
    textposition="outside",
    hovertemplate=(
        "<b>%{y}</b><br>"
        "Clientes: %{x}<br>"
        "Percentual dos cancelados: %{customdata[0]:.1f}%"
        "<extra></extra>"
    ),
)

fig.update_layout(
    title_x=0.5,
    coloraxis_showscale=False,
    xaxis=dict(showgrid=True, gridcolor="#E5E7EB"),
    yaxis=dict(showgrid=False),
    margin=dict(l=30, r=45, t=70, b=40),
)

fig.show()
```

```python
colunas_servicos = [
    "ServicoTelefonico",
    "MultiplasLinhas",
    "ServicoInternet",
    "SegurancaOnline",
    "BackupOnline",
    "ProtecaoDispositivo",
    "SuporteTecnico",
    "StreamingTV",
    "StreamingFilmes",
]

nomes_servicos = {
    "ServicoTelefonico": "Telefone",
    "MultiplasLinhas": "Múltiplas linhas",
    "ServicoInternet": "Internet",
    "SegurancaOnline": "Segurança online",
    "BackupOnline": "Backup online",
    "ProtecaoDispositivo": "Proteção do dispositivo",
    "SuporteTecnico": "Suporte técnico",
    "StreamingTV": "Streaming de TV",
    "StreamingFilmes": "Streaming de filmes",
}

# Seleciona clientes que cancelaram
cancelou = (
    df_gold["Cancelamento"]
    .astype("string")
    .str.strip()
    .str.casefold()
    .isin(["sim", "yes", "true", "1"])
)
clientes_cancelados = df_gold.loc[cancelou].copy()

if clientes_cancelados.empty:
    raise ValueError("Não foram encontrados clientes cancelados em df_gold.")

positivos = {"sim", "yes", "true", "1"}
negativos = {
    "não", "nao", "no", "false", "0", "",
    "none", "no internet service", "no phone service",
}

# Identifica os serviços ativos para cada cliente cancelado
servicos_ativos = pd.DataFrame(index=clientes_cancelados.index)

for coluna in colunas_servicos:
    valores = (
        clientes_cancelados[coluna]
        .astype("string")
        .str.strip()
        .str.casefold()
    )

    if coluna == "ServicoInternet":
        # Valores como DSL e Fiber optic indicam serviço de internet.
        servicos_ativos[coluna] = valores.notna() & ~valores.isin(negativos)
    else:
        servicos_ativos[coluna] = valores.isin(positivos)

# Resume a quantidade e os nomes dos serviços de cada cliente
analise_servicos = pd.DataFrame(index=clientes_cancelados.index)
analise_servicos["QuantidadeServicos"] = servicos_ativos.sum(axis=1)
analise_servicos["MaisDeUmServico"] = analise_servicos["QuantidadeServicos"] > 1
analise_servicos["Servicos"] = servicos_ativos.apply(
    lambda linha: [
        nomes_servicos[coluna]
        for coluna in colunas_servicos
        if linha[coluna]
    ],
    axis=1,
)

# Gráfico 1: clientes com até um serviço versus mais de um
analise_servicos["Grupo"] = analise_servicos["MaisDeUmServico"].map({
    False: "Até 1 serviço",
    True: "Mais de 1 serviço",
})

resumo = (
    analise_servicos["Grupo"]
    .value_counts()
    .reindex(["Até 1 serviço", "Mais de 1 serviço"], fill_value=0)
    .rename_axis("Grupo")
    .reset_index(name="Clientes")
)

fig_resumo = px.pie(
    resumo,
    names="Grupo",
    values="Clientes",
    hole=0.55,
    color="Grupo",
    color_discrete_map={
        "Até 1 serviço": "#93C5FD",
        "Mais de 1 serviço": "#2563EB",
    },
    title="Clientes cancelados por quantidade de serviços",
    template="plotly_white",
)

fig_resumo.update_traces(
    textinfo="label+percent",
    hovertemplate="<b>%{label}</b><br>Clientes: %{value}<extra></extra>",
)
fig_resumo.update_layout(title_x=0.5, legend_title_text="")
fig_resumo.show()

# Gráfico 2: combinações mais comuns entre quem tinha mais de um serviço
combinacoes = (
    analise_servicos.loc[analise_servicos["MaisDeUmServico"], "Servicos"]
    .apply(lambda lista: " + ".join(lista))
    .value_counts()
    .head(10)
    .sort_values()
    .rename_axis("Combinação de serviços")
    .reset_index(name="Clientes")
)

if not combinacoes.empty:
    fig_combinacoes = px.bar(
        combinacoes,
        x="Clientes",
        y="Combinação de serviços",
        orientation="h",
        text="Clientes",
        color="Clientes",
        color_continuous_scale="plasma",
        title="Combinações de serviços mais comuns",
        labels={"Clientes": "Clientes cancelados"},
        template="plotly_white",
    )

    fig_combinacoes.update_traces(
        textposition="outside",
        hovertemplate="<b>%{y}</b><br>Clientes: %{x}<extra></extra>",
    )
    fig_combinacoes.update_layout(
        title_x=0.5,
        coloraxis_showscale=False,
        yaxis_title="",
        xaxis=dict(showgrid=True, gridcolor="#E5E7EB"),
        margin=dict(l=30, r=40, t=70, b=40),
    )
    fig_combinacoes.show()
else:
    print("Não há clientes cancelados com mais de um serviço.")
```

```python

from sklearn.feature_extraction.text import CountVectorizer


# Seleciona os comentários dos clientes
comentarios = df_gold["FeedbackCliente"].dropna().astype(str)

# Palavras extras que não queremos mostrar no gráfico
palavras_ignoradas = [
    "customer",
    "customers",
    "company",
    "provider",
    "overall",
    "experience",
    "com",
    "a",
    "e"
]

# Separa as palavras e remove palavras comuns em inglês
vetorizador = CountVectorizer(stop_words="english")
matriz_palavras = vetorizador.fit_transform(comentarios)

# Conta quantas vezes cada palavra aparece
frequencias = matriz_palavras.sum(axis=0).A1
palavras = vetorizador.get_feature_names_out()

# Cria uma tabela com as palavras e suas frequências
df_palavras = pd.DataFrame({
    "Palavra": palavras,
    "Frequência": frequencias
})

# Remove as palavras extras da lista de exclusão
df_palavras = df_palavras[
    ~df_palavras["Palavra"].isin(palavras_ignoradas)
]

# Seleciona as 20 palavras mais frequentes
top_20 = (
    df_palavras
    .nlargest(20, "Frequência")
    .sort_values("Frequência", ascending=True)
)

# Cria o gráfico de barras
fig = px.bar(
    top_20,
    x="Frequência",
    y="Palavra",
    orientation="h",
    text="Frequência",
    color="Frequência",
    color_continuous_scale=["#BFDBFE", "#1D4ED8"],
    title="20 palavras mais frequentes nos comentários",
    labels={
        "Palavra": "Palavra",
        "Frequência": "Quantidade de aparições"
    },
    template="plotly_white"
)

fig.update_traces(textposition="outside")

fig.update_layout(
    title_x=0.5,
    coloraxis_showscale=False,
    xaxis_title="Quantidade de aparições",
    yaxis_title=""
)

fig.show()
```

```python
from wordcloud import STOPWORDS, WordCloud
import matplotlib.pyplot as plt

# Palavras que não queremos exibir
palavras_ignoradas = set(STOPWORDS)
palavras_ignoradas.update(
    [
        "customer",
        "customers",
        "company",
        "provider",
        "overall",
        "experience",
        "also",
        "really",
    ]
)

# 1. Separa os comentários com base na coluna de Cancelamento
texto_nao_cancelou = " ".join(
    df_gold[df_gold["Cancelamento"] == "No"]["FeedbackCliente"].dropna().astype(str)
)
texto_cancelou = " ".join(
    df_gold[df_gold["Cancelamento"] == "Yes"]["FeedbackCliente"].dropna().astype(str)
)

# 2. Gera a nuvem de palavras para quem NÃO cancelou (colormap azul)
nuvem_nao = WordCloud(
    width=700,
    height=700,
    background_color="#263CA2",
    stopwords=palavras_ignoradas,
    max_words=30,
    collocations=False,
    colormap="Blues",
    random_state=42,
).generate(texto_nao_cancelou)

# 3. Gera a nuvem de palavras para quem CANCELOU (colormap vermelho)
nuvem_sim = WordCloud(
    width=700,
    height=700,
    background_color="#E2616168",
    stopwords=palavras_ignoradas,
    max_words=30,
    collocations=False,
    colormap="Reds",
    random_state=42,
).generate(texto_cancelou)

# 4. Plota os dois gráficos lado a lado usando subplots
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

axes[0].imshow(nuvem_nao, interpolation="bilinear")
axes[0].axis("off")
axes[0].set_title(
    "Clientes que NÃO Cancelaram", fontsize=16, color="navy"
)

axes[1].imshow(nuvem_sim, interpolation="bilinear")
axes[1].axis("off")
axes[1].set_title(
    "Clientes que CANCELARAM", fontsize=16, color="darkred"
)

plt.tight_layout()
plt.show()
```
