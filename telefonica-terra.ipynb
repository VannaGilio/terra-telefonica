### Tradução e categorização das colunas

```python
#Tradução dos nomes das colunas
df.rename(columns={
    "customerID": "id_cliente",
    "gender": "Genero",
    "SeniorCitizen": "Senioridade",
    "Partner": "Parceiro",
    "Dependents": "Dependentes",
    "tenure": "DuracaoContrato",
    "PhoneService": "ServicoTelefonico",
    "MultipleLines": "MultiplasLinhas",
    "InternetService": "ServicoInternet",
    "OnlineSecurity": "SegurancaOnline",
    "OnlineBackup": "BackupOnline",
    "DeviceProtection": "ProtecaoDispositivo",
    "TechSupport": "SuporteTecnico",
    "StreamingTV": "StreamingTV",
    "StreamingMovies": "StreamingFilmes",
    "Contract": "Contrato",
    "PaperlessBilling": "FaturaEletronica",
    "PaymentMethod": "MetodoPagamento",
    "MonthlyCharges": "CobrancaMensal",
    "TotalCharges": "CobrancaTotal",
    "Churn": "Cancelamento",
    "CustomerFeedback": "FeedbackCliente",
    "MonthlyIncome": "RendaMensal",
}, inplace=True)
```

```python
df.head()
```

```python
# Verificando os tipos de dados das colunas
df.dtypes
```

```python
# Categorização correta das colunas
df = df.astype({
    "Genero": "category",
    "Senioridade": "category",
    "Parceiro": "category",
    "Dependentes": "category",
    "ServicoTelefonico": "category",
    "MultiplasLinhas": "category",
    "ServicoInternet": "category",
    "SegurancaOnline": "category",
    "BackupOnline": "category",
    "ProtecaoDispositivo": "category",
    "SuporteTecnico": "category",
    "StreamingTV": "category",
    "StreamingFilmes": "category",
    "Contrato": "category",
    "FaturaEletronica": "category",
    "MetodoPagamento": "category",
    "Cancelamento": "category",
})
```

```python
# Análise da coluna "DuracaoContrato"
df["DuracaoContrato"].value_counts(dropna=False)
```

```python
# Conversão da coluna "DuracaoContrato" para o tipo numérico, tratando valores não numéricos como NaN
df["DuracaoContrato"] = (
    pd.to_numeric(df["DuracaoContrato"], errors="coerce")
    .astype("Int64")
)

# Conversão das colunas restantes para os tipos de dados apropriados
df = df.astype({
    "id_cliente": "string",
    "CobrancaMensal": "float64",
    "CobrancaTotal": "float64",
    "FeedbackCliente": "string",
    "Senioridade": "Int64",
    "RendaMensal": "Int64",
    "DuracaoContrato": "Int64",
})
```
